```python
num=[334,571002,758,538,59,43,79]
even=[]
odd=[]
for x in num:
    if(x%2==0):
        even.append(x)
    else:
        odd.append(x)

print(f"The odd numbers are {odd}")
print(f"The even numbers are {even}")
```
